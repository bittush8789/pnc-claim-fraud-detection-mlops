# Exploratory Data Analysis (EDA) - P&C Insurance Claim Fraud Detection

This notebook explores the dataset to understand fraud patterns, distributions, and correlation among features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

## 1. Load the Dataset

```python
df = pd.read_csv('../insurance_claims.csv')
df.head()
```

## 2. Basic Dataset Info

```python
print(f"Shape: {df.shape}")
df.info()
```

```python
df.describe()
```

## 3. Exploratory Data Visualizations

### A. Target Variable Distribution (`fraud_reported`)

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='fraud_reported', data=df, palette='Set2')
plt.title('Distribution of Fraud reported')
plt.show()
```

### B. Claim Amount Distribution

```python
sns.histplot(data=df, x='claim_amount', hue='fraud_reported', kde=True, bins=30, multiple='stack')
plt.title('Claim Amount Distribution by Fraud Status')
plt.show()
```

### C. Policy Type Analysis

```python
sns.countplot(data=df, x='policy_type', hue='fraud_reported', palette='viridis')
plt.title('Fraud Reported by Policy Type')
plt.show()
```

### D. Incident Severity Analysis

```python
sns.countplot(data=df, x='incident_severity', hue='fraud_reported', order=['Minor', 'Major', 'Total Loss'])
plt.title('Fraud Reported by Incident Severity')
plt.show()
```

### E. Outlier Analysis (Boxplot)

```python
sns.boxplot(data=df, x='fraud_reported', y='claim_amount')
plt.title('Claim Amount Boxplot by Fraud Status')
plt.show()
```

### F. Correlation Heatmap

```python
plt.figure(figsize=(10, 8))
numeric_cols = df.select_dtypes(include=[np.number]).columns
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

### G. Fraud Trends by State

```python
sns.barplot(data=df, x='state', y='fraud_reported', errorbar=None, palette='crest')
plt.title('Fraud Rate by State')
plt.show()
```

### H. Fraud Trends by Claim Type

```python
sns.barplot(data=df, x='claim_type', y='fraud_reported', errorbar=None)
plt.title('Fraud Rate by Claim Type')
plt.show()
```
